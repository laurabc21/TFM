```python
import re
import pandas as pd

def extract_section(text):
    if pd.isna(text):
        return text

    # remove everything starting from "Okay, let me try"
    cleaned = re.split(r"\n?\s*Okay, let me try.*", text, flags=re.IGNORECASE | re.DOTALL)[0]

    return cleaned.strip()


df_spanish['translation'] = df_output_es['translation'].apply(extract_section)
```

```python
import json
```

```python
df_output_ca = pd.read_json('/content/output_CAT_baseline_qwen', lines=True)

df_output_ca.columns
```

```python
df_catalan['translation'] = df_output_ca['translation']
```

```python
import pandas as pd

def extract_until_last_dot_newline(text):
    if pd.isna(text):
        return text

    idx = text.rfind(".\n")

    if idx != -1:
        return text[:idx + 1].strip()

    return text.strip()


df_catalan['translation'] = df_output_ca['translation'].apply(extract_until_last_dot_newline)
```

```python
# Convert to Excel
df_catalan.to_excel('Qwen_CAT_baseline.xlsx', index=False)
```
