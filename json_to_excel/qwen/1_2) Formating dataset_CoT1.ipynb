```python
df_spanish['translation'] = df_output_es['translation']
```

```python
import re
import pandas as pd

def extract_section(text):
    if pd.isna(text):
        return text

    match = re.search(
        r"",
        text,
        re.IGNORECASE | re.DOTALL
    )

    if match:
        return match.group(1).strip()

    return text.strip()


df_spanish['translation'] = df_output_es['translation'].apply(extract_section)
```

```python
import re
import pandas as pd

def extract_texto_a_traducir(text):
    if pd.isna(text):
        return text

    match = re.search(
        r"Texto a traducir:\s*(.*?)(?=\n\s*Texto posterior:|\n\s*Texto anterior:|$)",
        text,
        re.DOTALL | re.IGNORECASE
    )

    if match:
        return match.group(1).strip()

    return text.strip()


df_spanish['translation'] = df_output_es['translation'].apply(extract_texto_a_traducir)
```
