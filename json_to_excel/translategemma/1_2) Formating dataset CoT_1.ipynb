```python
df_output_es = pd.read_json('/content/output_CoT1_ES.jsonl', lines=True)

df_output_es.columns
```

```python
df_spanish['translation'] = df_output_es['translation']
```

```python
import re
import pandas as pd

def extract_translation_or_keep(text):
    if pd.isna(text):
        return text  # keep NaN

    match = re.search(r'Translation:\s*(.*)', text, flags=re.DOTALL)

    if match:
        return match.group(1).strip()

    return text  # if no "Traducción:", keep full original output

df_spanish['translation'] = df_output_es['translation'].apply(extract_translation_or_keep)
```

```python
import re
import pandas as pd

def extract_translation(text):
    if pd.isna(text):
        return text  # keep NaN as is

    match = re.search(r'Traducción:\s*(.*)', text, flags=re.DOTALL)

    if match:
        return match.group(1).strip()

    return text  # keep original if no "Traducción:" found

df_catalan["translation"] = df_catalan["translation"].apply(extract_translation)
```
