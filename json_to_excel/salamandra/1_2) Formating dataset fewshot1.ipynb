```python
import pandas as pd
import json

def extract_translation(text):
    if pd.isna(text):
        return text

    try:
        data = json.loads(text)
        return data.get("translation", "")
    except:
        return text

df_spanish['translation'] = df_output_es['translation'].apply(extract_translation)
```

```python
df_output_ca = pd.read_json('/content/output_CAT_fewshot1_salamandra.jsonl', lines=True)

df_output_ca.columns
```

```python
df_catalan['translation'] = df_output_ca['translation']
```

```python
import re
import pandas as pd

def extract_texto_a_traducir(text):
    if pd.isna(text):
        return text

    match = re.search(
        r"Texto a traducir:\s*(.*?)(?=\n\s*Texto posterior:|\n\s*Texto anterior:|$)",
        text,
        re.DOTALL | re.IGNORECASE
    )

    if match:
        return match.group(1).strip()

    return text.strip()


df_catalan['translation'] = df_output_ca['translation'].apply(extract_texto_a_traducir)
```

```python
# Convert to Excel
df_catalan.to_excel('Salamandra_CAT_fewshot1.xlsx', index=False)
```
